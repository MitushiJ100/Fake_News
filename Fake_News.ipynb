```python
import numpy as np
import pandas as pd
import itertools
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
#Read the data
df=pd.read_csv('/Users/mitushijain/PycharmProjects/Fake_News/news.csv')

#Get shape and head
df.shape
df.head()

# The response variable: text
# the predictor variable: label
```

```python
# divide the dataset into test and train

x_data = df['text']
y_data = df['label']
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.2, random_state = 0)
```

```python
tfidf_vectorizer=TfidfVectorizer(stop_words='english', max_df=0.7) # initialize TfidfVectorizer (convert words to numbers)

# fit and transform the test and train dataset (both at the same time)
x_train_tfidf = tfidf_vectorizer.fit_transform(x_train)
x_test_tfidf = tfidf_vectorizer.transform(x_test)

pac_classifier = PassiveAggressiveClassifier(max_iter=50)

fit_pac = pac_classifier.fit(x_train_tfidf, y_train)

y_predict = pac_classifier.predict(x_test_tfidf)

score=accuracy_score(y_test,y_predict) # calculate accuracy
print(score)

confusion_matrix(y_test,y_predict, labels=['FAKE','REAL'])
```
